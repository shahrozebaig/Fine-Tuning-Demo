# **Fine-Tuning**



**1. What fine-tuning does is that it involves utilizing a pre-existing machine learning mode**

**2. One which already knows general concepts—and enhancing it further through the training process specifically for performing tasks in a certain domain.**

**3. Rather than designing a brand new system, fine-tuning involves giving it even more examples relevant to its new domain of operation, such as responses for customer support, information for medicine, or even programming syntax.**





# **Why Fine-Tuning is Needed**

**A regular AI model produces general outputs; however, when we require precise outputs, then fine-tuning is required. By providing specialized training data, we train the model on understanding a specific job. It can be medical guidance, customer service, or software development. In doing so, we ensure the model provides precise responses for our intended purpose.**

| ----------- | -------------------- | --------------------------- |
| Knowledge   | General knowledge    | Specific knowledge          |
| Accuracy    | Normal / average     | Higher accuracy             |
| Answers     | Generic answers      | More relevant answers       |
| Use case    | Works for many tasks | Best for one specific task  |
| Example     | General chatbot      | Medical or coding assistant |
| Performance | Good                 | Better for specific needs   |


# Full fine-tuning



**1. When full fine-tuning occurs, the model is trained again from scratch on new data, with no freezing of any portion of the model.**

**2. The process involves updating the model's weights and layers with the training data to solve the new task.**

**3. The approach enables the model to gain knowledge about the task in detail.**

**4. The technique also demands more time and resources because it trains the entire model.**

**5. It needs a large amount of data to train properly and avoid mistakes.**

**6. Although full fine-tuning is costly compared to other approaches, it is more efficient in solving certain tasks.**




# LoRA (Low-Rank Adaptation)

**1. Is a method where only small additional layers are trained instead of the whole model.**

**2. The original model weights are kept frozen and not changed during training.**

**3. It reduces memory usage and makes training much faster and gives good performance for specific tasks.**

**4. LoRA works by adding small trainable parameters that adapt the model to new data.**



# Difference Between Full Fine-Tuning and LoRA


| Aspect           | Full Fine-Tuning                   | LoRA (Low-Rank Adaptation)                      |
| ---------------- | ---------------------------------- | ----------------------------------------------- |
| Training         | The entire model is trained again  | Only small added layers are trained             |
| Weights          | All weights are updated            | Original weights are frozen                     |
| Model Change     | Fully modifies the model           | Adds small changes on top of model              |
| Memory Usage     | High memory required               | Very low memory required                        |
| Speed            | Slow training process              | Faster training process                         |
| Compute Power    | Needs powerful hardware            | Works on limited resources                      |
| Cost             | Expensive                          | Cost-effective                                  |
| Accuracy         | Very high (best for complex tasks) | Good accuracy (slightly lower than full tuning) |
| Data Requirement | Needs large dataset                | Works with smaller datasets                     |
| Flexibility      | Less flexible (one model per task) | More flexible (can switch adapters easily)      |


# QLoRA (Quantized LoRA)

**1. QLoRA (Quantized LoRA) is a method used to fine-tune AI models efficiently.**

**2. The original model is compressed to low precision (like 4-bit) to save memory.**

**3. It uses less memory, less cost, and less compute power.**

**4. It gives performance close to full fine-tuning, but with much less cost.**

# PEFT (Parameter-Efficient Fine-Tuning)

**PEFT (Parameter-Efficient Fine-Tuning) is a way of training AI models where instead of changing the whole model, only a small part of it is trained while the rest stays the same. This makes the process faster, cheaper, and easier to run on normal computers. Methods like LoRA and QLoRA are examples of PEFT, where small extra layers are added and trained without touching the original model. Compared to full fine-tuning, which updates everything and needs more data and power, PEFT is a smarter and more efficient approach that still gives good performance for specific tasks.**

# Steps for Fine-Tuning

**Choose a pre-trained model**

* Start with a model that already knows general knowledge.

**Prepare your dataset**
* Collect and clean data related to your task (like medical, chatbot, coding, etc.).

**Format the data**
* Convert your data into the correct input-output format for training.

**Select fine-tuning method**
* Choose how you want to train (Full Fine-Tuning, LoRA, QLoRA, etc.).

**Train the model**
* Feed the data to the model and let it learn by updating weights.

**Evaluate the model**
* Test the model to check how well it performs.

**Save and use the model**
* Save the trained model and use it for your application.

# Requirements for Fine-Tuning

| Requirement         | Description                       | Example                                          |
| ------------------- | --------------------------------- | ------------------------------------------------ |
| Pre-trained Model   | Base model with general knowledge | GPT-like model or BERT                           |
| Dataset             | Task-specific data                | 10k–100k examples (chat, medical, etc.)          |
| Data Format         | Input → Output format             | “Question → Answer” pairs                        |
| Computing Resources | Hardware needed for training      | CPU works for small tasks, but GPU is better     |
| Memory (RAM/VRAM)   | Memory to load model              | 8–16 GB RAM (LoRA), 16–48 GB VRAM for big models |
| Fine-Tuning Method  | Type of tuning used               | Full tuning / LoRA / QLoRA                       |
| Tools/Frameworks    | Software for training             | PyTorch, TensorFlow, Hugging Face                |
